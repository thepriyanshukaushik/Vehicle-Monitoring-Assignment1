Drawing Shapes

```python
import cv2
```

```python
import numpy as np
```

```python
blank = np.zeros((500,500,3), dtype = 'uint8')
```

```python
from matplotlib import pyplot as plt
```

```python
plt.imshow(blank)
```

```python
# Changing the colours with slicing
blank[200:300, 300:400] = 0,255,2
```

```python
blank
```

```python
# thickness can be number, cv2.FILLED or thickness = -1 as well
# cv2.rectangle(image, start_point, end_point, color, thickness)
rect = cv2.rectangle(blank, (0,0), (250,250), (0,255,2), thickness = cv2.FILLED)
```

```python
plt.imshow(rect)
```

```python
bz = np.zeros((500,500,3), dtype = 'uint8')
```

```python
# image and center coords are passed //2 divided by divided by 2 for y and x axis
# cv2.circle(image, center, radius, color, thickness)
cir = cir = cv2.circle(bz, (bz.shape[1]//2, bz.shape[0]//2), 40, (0,0,255), thickness = 5) 
```

```python
plt.imshow(cir)
```

```python
lin = cv2.line(bz, (250,300), (100,300), (255,55,255), thickness = 10)
```

```python
plt.imshow(lin)
```

```python
bn = np.zeros((500,500,3), dtype= 'uint8')
```

```python
# Draw an image 

text = cv2.putText(bn, "Hello Priyanshu", (100, 100), cv2.FONT_HERSHEY_TRIPLEX, 1.2, (0,255,12), thickness = 3)
```

```python
plt.imshow(text)
```
