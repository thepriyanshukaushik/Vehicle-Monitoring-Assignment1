```python
import cv2
```

```python
img = cv2.imread('wall.jpg')
```

```python
cv2.imshow("Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img.shape
```

```python
img = cv2.resize(img, (860, 560))
```

```python
cv2.imshow("Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img.shape
```

```python
# RGB to GrayScale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
cv2.imshow("GaryScale", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray.shape
```

```python
#Converting into hsv
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
cv2.imshow("HSV", hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Working in Videos --> Either pass the address path or the 0 to turn on camera
vid = cv2.VideoCapture(0)
```

```python
#Playing the Video by frame by frame run infinite times 
while True:
    try:
        res, frame = vid.read()
        cv2.imshow("Video", frame)
    except:
        pass
    # At q is pressed window close
    if cv2.waitKey(1) & 0xff==ord('q'):
        break;

cv2.destroyAllWindows()
```
